```python
null_perc=(df_EC2.isnull().sum()/len(df_EC2))*100
null_perc = null_perc[null_perc>0]
null_perc = null_perc.sort_values(ascending=False)
null_perc
```

```python
# To see the outliers for replacing the null values whether by mean or median
import matplotlib.pyplot as plt

cols = [
    'Age',
    'Payment_Method_Diversity',
    'Email_Open_Rate',
    'Days_Since_Last_Purchase',
    'Pages_Per_Session',
    'Session_Duration_Avg',
    'Product_Reviews_Written',
    'Discount_Usage_Rate',
    'Wishlist_Items',
    'Returns_Rate',
    'Mobile_App_Usage',
    'Credit_Balance',
    'Social_Media_Engagement_Score',
    'Customer_Service_Calls'
]

fig, axes = plt.subplots(4, 4, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(cols):
    axes[i].boxplot(df_EC2[col].dropna())
    axes[i].set_title(col, fontsize=10)
    axes[i].tick_params(axis='x', labelbottom=False)

# Hide unused subplots (16 slots, 14 columns)
for j in range(len(cols), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.title("Outliers")
plt.show()
```

```python
# here all cols having outliers so we can replace null values with its median values.
```

```python
# Remove rows which have age of year '150' and '200'
df_EC2 = df_EC2[~df_EC2['Age'].isin([150, 200])].copy()
```

```python
# Find median of all remaining cols that have null values
for col in df_EC2.columns:
    if df_EC2[col].isnull().sum() > 0:
        print(f"{col} --> Median = {df_EC2[col].median()}")
```

```python
# Filling null values with its median value in all Null value col

for col in df_EC2.select_dtypes(include=['int64', 'float64']).columns:
    if df_EC2[col].isnull().sum() > 0:
      med_value=df_EC2[col].median()
      df_EC2[col]=df_EC2[col].fillna(med_value)
```
